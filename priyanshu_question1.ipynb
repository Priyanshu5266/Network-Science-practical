1. Manual Implementation: Determine the total number of nodes (N) and manually count the
degree (k_i) of every node. Calculate degree centrality using:

```python
print ("karate club nodes=",karate.number_of_nodes())
print ("karate club edges=",karate.number_of_edges())
```

```python
print("dolphins nodes:=",dolphins.number_of_nodes())
print("dolphins edges:=",dolphins.number_of_edges())
```

```python
def manual_degree(G):

    degree = {}

    for node in G.nodes():
        count = 0

        for neighbor in G.neighbors(node):
            count = count + 1

        degree[node] = count

    return degree
```

```python
karate_degree = manual_degree(karate)

dolphin_degree = manual_degree(dolphins)

print("Karate degree calculated.")
print("Dolphin degree calculated.")
```

```python
def manual_degree_centrality(G, degree):
  N=G.number_of_nodes()
  centrality={}
  for node in G.nodes():
    centrality[node]=degree[node]/(N-1)
  return centrality
```

```python
karate_centrality= manual_degree_centrality(karate,karate_degree)
dolphin_centrality= manual_degree_centrality(dolphins,dolphin_degree)
print("Karate centrality and dolphins calculated.")
```

```python
print("KARATE CLUB NETWORK")
print("Node\tDegree\tCentrality")

for node in karate.nodes():
    print(
        node,"\t",karate_degree[node],"\t",(karate_centrality[node])
    )
```

```python
print("DOLPHIN NETWORK")
print("Node\tDegree\tCentrality")

for node in dolphins.nodes():
    print(
        node,"\t",dolphin_degree[node],"\t",(dolphin_centrality[node])
    )
```

2. Ranking Analysis: Display the degree and manually calculated degree centrality for all nodes,
and identify the top 5 most connected individuals in each network.

```python
karate_top5= sorted(
    karate_degree.items(),
    key=lambda x:x[1],
    reverse=True
)[:5]
print("TOP 5 KARATE CLUB NODES")

for node, degree in karate_top5:
    print(
        "Node:", node,
        "Degree:", degree,
        "Centrality:", (karate_centrality[node])
    )
```

```python
dolphin_top5= sorted(
    dolphin_degree.items(),
    key=lambda x:x[1],
    reverse=True
)[:5]
print("TOP 5 DOLPHIN NODES")

for node, degree in dolphin_top5:
  print("Node:", node,
        "Degree:", degree,
        "Centrality:", dolphin_centrality[node])
```

3. Visual Analysis: Visualize both networks with node sizes proportional to the manually calculated
degree centrality and identify the prominent individuals.

```python
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(karate, seed=3)

node_sizes = []

for node in karate.nodes():
    node_sizes.append(karate_centrality[node] * 3000)

nx.draw(
    karate,
    pos,
    with_labels=True,
    node_size=node_sizes
)

plt.title("Karate Club - Degree Centrality")
plt.show()
```

```python
plt.figure(figsize=(12, 10))

pos = nx.spring_layout(dolphins, seed=3)

node_sizes = []

for node in dolphins.nodes():
    node_sizes.append(dolphin_centrality[node] * 3000)

nx.draw(
    dolphins,
    pos,
    with_labels=True,
    node_size=node_sizes
)

plt.title("Dolphin Social Network - Degree Centrality")
plt.show()
```
