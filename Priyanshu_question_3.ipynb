**Question 3: Consider an Airline Travel Reachability Network where cities are represented as nodes and a
directed edge A→B indicates that city B can be reached from city A within a specified travel-time
threshold. The network is asymmetric, meaning A→B does not necessarily imply B→A. Using Python
and NetworkX, perform the following tasks:**

```python
import networkx as nx
```

```python
import pandas as pd
```

```python
from collections import deque
```

```python
import gzip
```

1. Construct the Network: Load the airline travel dataset, create a directed graph, and determine
the number of nodes and edges.

```python
G= nx.DiGraph()
with gzip.open("reachability.txt.gz", "rt") as file:

  for line in file:
    if not line.startswith("#"):
      data = line.split()
      G.add_edge(int(data[0]), int(data[1]))

print("number of nodes:", G.number_of_nodes())
print("number of eedges: ", G.number_of_edges())
```

```python
meta= pd.read_csv("reachability-meta.csv.gz")
print(meta.head())
print(meta.columns)
```

```python
city_names = dict(zip(meta["node_id"], meta["name"]))

print(city_names[0])
print(city_names[1])
```


2. Calculate Centrality Measures: For each city, calculate the in-degree, out-degree, in-degree
centrality, out-degree centrality, and closeness centrality. Implement degree centrality manually
using:




```python
for node in G.nodes():
    in_degree = G.in_degree(node)
    out_degree = G.out_degree(node)

    print(city_names[node],"In-degree:", in_degree,
          "Out-degree:", out_degree)
```

```python
N = G.number_of_nodes()
for node in G.nodes():
  in_centrality = G.in_degree(node) / (N - 1)
  out_centrality = G.out_degree(node) / (N - 1)

  print(city_names[node],"In-degree centrality:", in_centrality,
        "Out-degree centrality:", out_centrality)
```

Implement Closeness from Scratch: Use Breadth-First Search (BFS) to calculate shortest-path
distances and closeness centrality without using nx.closeness_centrality(), nx.shortest_path(), or
nx.shortest_path_length().

```python
def bfs_distances(G, start):
  distances = {start:0}
  queue = deque([start])

  while queue:
    current = queue.popleft()
    for neighbor in G.neighbors(current):
      if neighbor not in distances:
        distances[neighbor] = distances[current] + 1
        queue.append(neighbor)
  return distances
```

```python
closeness={}
N= G.number_of_nodes()

for node in G.nodes():
  distances = bfs_distances(G,node)
  total_distance = sum(distances.values())
  closeness[node] = (N-1)/ total_distance

print("closeness calculation done")
```

```python
results = []

for node in G.nodes():
    results.append([
        city_names[node],
        G.in_degree(node),
        G.out_degree(node),
        G.in_degree(node) / (N - 1),
        G.out_degree(node) / (N - 1),
        closeness[node]
    ])

df = pd.DataFrame(results, columns=[
    "City",
    "In-Degree",
    "Out-Degree",
    "In-Degree Centrality",
    "Out-Degree Centrality",
    "Closeness Centrality"
])

print(df.head())
```
